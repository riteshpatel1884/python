```bash
Logging means recording what's happening inside your application while it runs.

It's similar to print(), but much more powerful and suitable for real applications.

For example, instead of:

print("User logged in")
print("Fetching data")
print("API failed")

use:

import logging
logging.info("User logged in")
logging.info("Fetching data")
logging.error("API failed")

Log levels
| Level      | Use                                            |
| ---------- | ---------------------------------------------- |
| `DEBUG`    | Detailed information for debugging             |
| `INFO`     | Normal application events                      |
| `WARNING`  | Something unexpected but not necessarily fatal |
| `ERROR`    | Something failed                               |
| `CRITICAL` | Serious failure                                |


```bash

Logging vs print()

print(): Good for quick debugging.

But logging gives you:

Severity levels
Timestamps
File output
Better formatting
Exception information
Configuration
Different behavior for development/production

For production applications, logging is much better than scattered print() statements

````python
# Logging exceptions:  This is extremely useful.

import logging
try:
    result = 10 / 0

except ZeroDivisionError:
    logging.exception("Failed to calculate result")
````

````python
# Logging to a file: Instead of displaying logs in the terminal:

import logging
logging.basicConfig(
    filename="app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Something looks suspicious")
````
