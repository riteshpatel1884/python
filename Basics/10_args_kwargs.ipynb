- *args - *args allows a function to accept any number of positional arguments.
- Positional arguments are arguments passed to a function based on their position/order.

```python
def add(*args):
    print(args)

add(10, 20, 30, 40)
```

```python
def student(name, age, city):
    print(name)
    print(age)
    print(city)

student("Ritesh", 21, "Delhi")

# So these are positional arguments because their position determines where they go.
```

```python
# Compare with keyword arguments
def student(name, age, city):
    print(name)
    print(age)
    print(city)

student(name="Ritesh", age=21, city="Delhi")

# Here you are explicitly telling Python:
# name = "Ritesh"
# age  = 21
# city = "Delhi"
# so these are keyword arguments.
```

- **kwargs - **kwargs allows a function to accept any number of keyword arguments.
- Keyword arguments are arguments passed using name=value.

```python
def student(**kwargs):
    print(kwargs)

student(name="Ritesh", age=21, branch="CSE")
```

```python
# You can access them like a dictionary:
def student(**kwargs):
    print(kwargs["name"])
    print(kwargs["age"])

student(name="Ritesh", age=21)
```

```python
# Why use **kwargs?
def student(name, age, city):
    print(name, age, city)

student("Ritesh", 22, "Delhi")
# Without **kwargs, You have to define every parameter. But what if you also want branch or college. You would have to change the function. 
# def student(name, age, city, branch, college):
#     ...

# We use **kwargs when we don't know beforehand how many keyword arguments a function will receive. You can pass as many named arguments as you want.
def student(**kwargs):
    print(kwargs)

student(
    name="Ritesh",
    age=22,
    city="Delhi",
    branch="CSE",
    college="KIET"
)
```

```python
# Using both together
def demo(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

demo(10, 20, 30, name="Ritesh", age=21)
```

| Syntax     | Collects             | Python type |
| ---------- | -------------------- | ----------- |
| `*args`    | Positional arguments | Tuple       |
| `**kwargs` | Keyword arguments    | Dictionary  |
