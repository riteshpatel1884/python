A generator is a simple way to create an iterator.

The key difference is:

A normal function returns a result and finishes.
A generator uses yield to return a value one at a time and remembers where it stopped.

A generator is a type of iterator, and yield provides an easy way to create one.

```python
def count():
    yield 1
    yield 2
    yield 3

numbers = count()  # numbers is a generator object.

print(next(numbers))  # 1
print(next(numbers))  # 2
print(next(numbers))  # 3
```

yield → give a value, pause the function, and resume it later.

```python
def count():
    print("Start")
    yield 1
    print("Second")
    yield 2
    print("Third")
    yield 3

numbers = count()
# the function doesn't actually run yet. It creates the generator.

print(next(numbers))  # Python starts executing:
```

The generator pauses at yield 1.

```python
print(next(numbers))
```

Then the next next() resumes again.

next() → Start → yield 1 → PAUSE


next() → resume → Second → yield 2 → PAUSE


next() → resume → Third → yield 3 → PAUSE


next() → resume → END → StopIteration

Generator vs normal function


```python
#Normal fctn: creates the entire list at once.
def numbers():
    return [1, 2, 3]

# Generator: It produces values one at a time.
def numbers():
    yield 1
    yield 2
    yield 3
```

```python
# Why are generators useful?

# The biggest reason is memory efficiency.

# Imagine:

def numbers():
    for i in range(1000000000):
        yield i

# You can do:

for n in numbers():
    print(n)

# It doesn't need to create a list containing 1 billion numbers.
# It generates:
# 0 → process
# 1 → process
# 2 → process
# 3 → process
...

# only when needed.


# This is extremely useful for:
# Large datasets
# Reading large files
# Data pipelines
# Streaming data
# AI/ML pipelines
# Processing large amounts of records
```

```python
# You previously had to write a whole class:
class Count:
    def __init__(self, max_value):
        self.current = 1
        self.max_value = max_value

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max_value:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration

# That's a lot of code.

# With a generator, the same idea becomes:

def count(max_value):
    current = 1

    while current <= max_value:
        yield current
        current += 1

numbers = count(3)

print(next(numbers))  # 1
print(next(numbers))  # 2
print(next(numbers))  # 3
```
