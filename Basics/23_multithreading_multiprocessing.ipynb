Multithreading = multiple threads inside one process

Multiprocessing = multiple processes running independently

```bash 
Multithreading

Multiple threads share the same process and memory. While one thread waits for a network response, another can do useful work.

Process:

├── Thread 1
├── Thread 2
├── Thread 3
└── Thread 4

Threads are particularly useful for I/O-bound tasks:
-> API 
-> Web scraping
-> File operations
-> Database queries
-> Network requests

`````python
import threading

def task(name):
    print(f"Running {name}")

t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))

t1.start()
t2.start()

t1.join()
t2.join()
`````

```bash
Multiprocessing creates separate Python processes.

Process 1 → CPU Core 1
Process 2 → CPU Core 2
Process 3 → CPU Core 3
Process 4 → CPU Core 4


Each process has its own memory space.
Multiprocessing is useful for CPU-bound tasks:
-> Heavy mathematical 
-> Image processing
-> Large data processing
-> CPU-heavy ML preprocessing
-> Computational simulations

`````python
from multiprocessing import Process

def task(name):
    print(f"Running {name}")

p1 = Process(target=task, args=("Task 1",))
p2 = Process(target=task, args=("Task 2",))

p1.start()
p2.start()

p1.join()
p2.join()
`````

```bash
GIL: CPython has the Global Interpreter Lock (GIL), which means generally only one thread at a time can execute Python bytecode within a process.

So if you have:
# CPU-heavy
for i in range(10**9):
    result += i

using multiple Python threads usually won't give you true CPU-parallel execution.
That's where multiprocessing can help:

Multithreading
Process
 ├── Thread 1 ──┐
 ├── Thread 2 ──┤ → GIL limits Python bytecode execution
 └── Thread 3 ──┘


Multiprocessing
Process 1 → CPU
Process 2 → CPU
Process 3 → CPU
Process 4 → CPU

→ True parallelism


| Feature         | Multithreading            | Multiprocessing  |
| --------------- | ------------------------- | ---------------- |
| Unit            | Thread                    | Process          |
| Memory          | Shared                    | Separate         |
| CPU parallelism | Limited by GIL in CPython | Yes              |
| Best for        | I/O-bound                 | CPU-bound        |
| Memory usage    | Lower                     | Higher           |
| Communication   | Easier                    | More complicated |
| Startup         | Faster                    | Slower           |
| Example         | API calls                 | Image processing |
