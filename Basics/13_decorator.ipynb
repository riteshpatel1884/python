A decorator is a function that allows you to add extra behavior to another function without changing its original code.

Think of it like putting a wrapper around a function.

Original function

      ↓

   Decorator
   
      ↓
Extra behavior + Original function

```python
def greet():
    print("Hello Ritesh")

greet()

```

Now imagine you want to add:

Before function → "Function started"
After function  → "Function finished"

You could modify greet(), but decorators let you do this without touching greet().

```python
def my_decorator(func):

    def wrapper():
        print("Before function")
        func()
        print("After function")

    return wrapper
```

Here func is the function we're decorating. And wrapper() is the new function that adds extra behavior.

```python
def greet():
    print("Hello Ritesh")

# Apply the decorator:
greet = my_decorator(greet)

# Then:
greet()

```
