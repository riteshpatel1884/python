```bash
Memory management is how Python allocates, uses, and releases memory while your program runs.
You don't normally have to manually allocate and free memory like in C/C++. Python handles most of this automatically.

Your Python program
       ↓
Creates objects
       ↓
Python allocates memory
       ↓
Objects are used
       ↓
Objects are no longer needed
       ↓
Python reclaims memory

```bash
Everything is an object

In Python:
x = 10
name = "Ritesh"
numbers = [1, 2, 3]

Python creates objects in memory.

Conceptually:
x ───────→ 10
name ────→ "Ritesh"
numbers ─→ [1, 2, 3]

The variables (x, name, numbers) are references to objects, rather than boxes that directly contain the values.

```bash
Reference Counting: One of Python's main memory-management mechanisms is reference counting.

Example:
x = [1, 2, 3]

The list has a reference from x.

If we do: y = x
now both x and y point to the same object:

x ──┐
    ├──→ [1, 2, 3]
y ──┘

If we then do: del x
the list isn't deleted because y still references it.

y ──→ [1, 2, 3]
Once there are no references left, Python can reclaim that object's memory.

Stack vs Heap

```bash

Stack: Used for things like function call information and local references.
def add():
    x = 10
    y = 20

Heap: Python objects are generally allocated in the heap.

Memory
│
├── Stack
│    └── references / call frames
│
└── Heap
     ├── integer object
     ├── list object
     ├── string object
     └── dictionary object

```bash
Python's Private Heap: CPython manages a private heap for Python objects.

When you write:
numbers = [1, 2, 3]

Python's memory manager handles allocating the memory required for the list and its objects. This is why you don't normally write things like:

malloc(...)
free(...)

as you would in C.

```bash
is vs == and Memory
This connects to object references.

a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True
print(a is b)  # False

== checks value equality.
is checks whether they are the same object.

a ──→ [1,2,3]
b ──→ [1,2,3]
Different objects
