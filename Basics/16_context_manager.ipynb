A context manager is used to properly set up and clean up resources.

The most common example is working with files.



```python
# Without a context manager
file = open("data.txt", "r")
data = file.read()
file.close()

# You have to remember to call file.close()

# If an exception occurs before close(), the file might remain open.
```

```python
# With a context manager Python provides the with statement:

with open("data.txt", "r") as file:
    data = file.read()

# as file is the variable
# Python automatically closes the file when the with block finishes.
```

```python
# How does with actually work?

# A context manager typically implements two special methods:
# __enter__()
# __exit__()

class MyContext:
    def __enter__(self):
        print("Entering")

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting")


with MyContext():
    print("Inside")

# with MyContext()
#        ↓
# __enter__()
#        ↓
# code inside with
#        ↓
# __exit__()
```

The context manager can automatically handle things like:

-> Opening the connection

-> Executing your code

-> Closing the connection

-> Cleaning up if an error occurs
