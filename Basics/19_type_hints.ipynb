Type hints are a way of telling Python and more importantly, developers and tools what type of data a variable, parameter, or return value is expected to have.

They use : and -> syntax.

```python
# Variable type hints
name: str = "Ritesh"
age: int = 21
salary: float = 50000.5
is_student: bool = True
```

```python
# Function parameter type hints
def greet(name: str):
    print(f"Hello {name}")
```

```python
# You can also specify the return type:

def add(a: int, b: int) -> int:
    return a + b

# Read it as add takes two integers and is expected to return an integer.
```

```python
# You can specify what a list contains:

numbers: list[int] = [1, 2, 3, 4]
names: list[str] = ["Ritesh", "Rahul", "Aman"]
```

```python
# Dictionaries
users: dict[str, int] = {
    "Ritesh": 21,
    "Rahul": 22
}
```

```python
Optional values: Sometimes a variable can contain either a value or None.


name: str | None = None
```
