An iterator is an object that lets you go through elements one at a time, instead of getting everything at once.

```python
numbers = [10, 20, 30, 40]

it = iter(numbers)

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
print(next(it))  # 40

# Every time you call next(), Python gives you the next value.
```

Iterable vs Iterator

Iterable → something you can loop over.
Ex: 
-> list
-> tuple
-> string
-> dictionary
-> set

A list is iterable, but it isn't itself an iterator. you can create an iterator from it: it = iter(numbers)


```python
# How for loop actually works
# When you write:

numbers = [10, 20, 30]
for n in numbers:
    print(n)

# Python is essentially doing:
it = iter(numbers)
while True:
    try:
        n = next(it)
        print(n)
    except StopIteration:
        break

# So for loops internally use iterators.
```

Creating your own iterator

A class becomes an iterator when it implements:
__iter__()
__next__()

```python
class Count:
    def __init__(self, max_value):
        self.current = 1
        self.max_value = max_value

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max_value:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration


numbers = Count(3)

print(next(numbers))  # 1
print(next(numbers))  # 2
print(next(numbers))  # 3
```

Why are iterators useful?

-> The biggest advantage is memory efficiency.

-> Suppose you need to process 10 million values. 

-> Creating a huge list: numbers = [1, 2, 3, ..., 10000000] stores all those values in memory.

-> An iterator can generate/process values one at a time:

Value 1 → process

Value 2 → process

Value 3 → process

...


-> This is why iterators are important when working with large datasets, files, generators, and data pipelines.

```python
#self

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Ritesh", 21)
student2 = Student("Rahul", 22)
```

There are now two separate objects.

student1

age  = 21

student2

When this happens:

student1 = Student("Ritesh", 21)

Python essentially calls:

Student.__init__(student1, "Ritesh", 21)

Therefore, inside __init__:

self = student1

So: self.name = name means: student1.name = "Ritesh"

And: self.age = age means: student1.age = 21
