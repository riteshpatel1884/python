Package management means installing, updating, removing, and managing the external libraries/packages your Python project depends on.

For example, if your AI project needs:

-> langchain

-> pandas

-> numpy

-> fastapi

-> qdrant-client

you need a way to install and manage them. That's package management.


````python
pip — the traditional package manager
````

Package vs Library vs Module

Module: A single .py file:

Package: A collection of Python modules organized together.

Library: A broader term for reusable code that you can use in your projects.
