map() → transform every element

filter() → select certain elements

reduce() → combine all elements into one result

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for x in numbers:
    squares.append(x * x)

print(squares)
```

```python
# With map():

squares = list(map(lambda x: x * x, numbers))
print(squares)
```

```python
#Map Without lambda:

numbers = [1, 2, 3, 4, 5]
def double(x):
    return x * x

result = list(map(double, numbers))
print(result)
```

```python
# With lambda:
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

```python
# filter Without lambda:

numbers = [1, 2, 3, 4, 5, 6]
def is_even(x):
    return x % 2 == 0

result = list(filter(is_even, numbers))
print(result)
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
result = reduce(lambda a, b: a + b, numbers)
print(result)
```

```python
# reduce Without lambda:
from functools import reduce

numbers = [1, 2, 3, 4, 5]
def add(a, b):
    return a + b

result = reduce(add, numbers)
print(result)
```
