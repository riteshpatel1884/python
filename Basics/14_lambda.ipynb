A lambda function is a small, anonymous function written in one line.

Syntax-> lambda arguments: expression

The result of the expression is automatically returned.

```python
# Instead of writing:

def square(x):
    return x * x

# You can write:

square = lambda x: x * x
print(square(5))
```

```python
# Multiple arguments 
add = lambda a, b: a + b

print(add(10, 20))
```

Why use lambda?

Lambda becomes particularly useful when you need a small function temporarily, especially with functions like:

map()

filter()

sorted()

```python
students = [
    ("Ritesh", 85),
    ("Rahul", 72),
    ("Aman", 91)
]

students.sort(key=lambda student: student[1])
print(students)
```

```python
students.sort(key=lambda student: student[1],reverse=True)
print(students)
```

```python
# Lambda with map()

numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x * x, numbers))
print(squares)

# numbers is simply the list of numbers you want to check.

# Without lambda:
# def square(x):
#     return x * x
# squares = list(map(square, numbers))
```

```python
# Lambda with filter()
numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)
```
