Comprehensions are a shorter and cleaner way to create lists or dictionaries using a loop, often with a condition.

```python
# List Comprehension

# Normal for loop
squares = []

for x in range(1, 6):
    squares.append(x * x)

print(squares)
```

```python
# Using list comprehension
# [expression for item in iterable]

squares = [x * x for x in range(1, 6)]
print(squares)
```

```python
# List Comprehension with Condition
# [expression for item in iterable if condition]

even = [x for x in range(1, 11) if x % 2 == 0]
print(even)
```

```python
names = ["ritesh", "rahul", "aman"]

upper_names = [name.upper() for name in names]
print(upper_names)
```

```python
# if-else in List Comprehension
# [expression_if_true if condition else expression_if_false for item in iterable]

result = ["Even" if x % 2 == 0 else "Odd" for x in range(1, 6)]

print(result)
```

Dictionary Comprehension

```python
# Normal loop
squares = {}

for x in range(1, 6):
    squares[x] = x * x

```

```python
# Dictionary comprehension
# {key_expression: value_expression for item in iterable}

squares = {x: x * x for x in range(1, 6)}
print(squares)
```

```python
# Dictionary Comprehension with Condition
numbers = [1, 2, 3, 4, 5, 6]
even_squares = {x: x*x for x in numbers if x % 2 == 0}
print(even_squares)
```

```python
students = {
    "Ritesh": 85,
    "Rahul": 45,
    "Aman": 72,
    "Priya": 91
}

# Get students who scored at least 70:
passed = {
    name: marks
    for name, marks in students.items()
    if marks >= 70
}

print(passed)
```

```python
# Nested List Comprehension
# Normal:

result = []

for i in range(3):
    for j in range(3):
        result.append((i, j))

print(result)
```

```python
# Comprehension:

result = [(i, j) for i in range(3) for j in range(3)]

print(result)
```
